Binary search

```python
def binary_search(arr, left, right, n):
    if left > right:
        return -1  # element not found

    mid = (left + right) // 2

    if arr[mid] == n:
        return mid
    elif n < arr[mid]:
        return binary_search(arr, left, mid - 1, n)
    else:
        return binary_search(arr, mid + 1, right, n)

a = [2, 4, 6, 7, 9, 11, 24, 37, 38]
print(binary_search(a, 0, len(a) - 1, 6))
```
