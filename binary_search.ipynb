Binary search
- TC : log n
- SC : 

```python
def binary_search(arr, left, right, n):
    if left > right:
        return -1  # element not found

    mid = (left + right) // 2

    if arr[mid] == n:
        return mid
    elif n < arr[mid]:
        return binary_search(arr, left, mid - 1, n)
    else:
        return binary_search(arr, mid + 1, right, n)

a = [2, 4, 6, 7, 9, 11, 24, 37, 38]
print(binary_search(a, 0, len(a) - 1, 6))
```

find lower bound of target

```python

a = [1,1,1,2,3,3,5,6,7,7,7,9,12,12,14]

low = 0
high = len(a)-1
lb = len(a)
target = 1

while low <= high:
	mid =(low + high)//2
	if a[mid] >= target:
		lb = mid
		high = mid-1
	else:
		low = mid + 1
		
print(lb)
```
