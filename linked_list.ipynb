You are given an array arr[] of integers. You have to construct a singly linked list from the elements of the arr[] and return the head of the linked list.

```python
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None

class Solution:
    def __init__(self):
        self.head = None
    def arrayToLinkedList(self,arr):
        if not arr:
            return None
        self.head = Node(arr[0])
        current = self.head
        for value in arr[1:]:
            current.next = Node(value)
            current = current.next
        return self.head
sol = Solution()
result = sol.arrayToLinkedList([1,2,3,4,5])
print(result.next.data)
```

Insert the node at the end of LL and keep the track of tail

```python
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None


class Solution:
    def __init__(self):
        self.tail = None  # ✅ Track last node

    def insertAtEnd(self, head, x):
        new_node = Node(x)

        # Case 1: Empty list
        if head is None:
            head = new_node
            self.tail = new_node  # first node becomes tail
            return head

        # Case 2: Non-empty list
        # If tail is not initialized (e.g., multiple test cases), find it once
        if self.tail is None:
            temp = head
            while temp.next:
                temp = temp.next
            self.tail = temp

        # Attach new node and move tail
        self.tail.next = new_node
        self.tail = new_node
        return head

sol = Solution()
result = sol.insertAtEnd(None, 10)
print(result.data)  # Should print 10
result = sol.insertAtEnd(result, 20)    
print(result.next.data)  # Should print 20
```
