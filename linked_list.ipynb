You are given an array arr[] of integers. You have to construct a singly linked list from the elements of the arr[] and return the head of the linked list.

```python
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None

class Solution:
    def __init__(self):
        self.head = None
    def arrayToLinkedList(self,arr):
        if not arr:
            return None
        self.head = Node(arr[0])
        current = self.head
        for value in arr[1:]:
            current.next = Node(value)
            current = current.next
        return self.head
sol = Solution()
result = sol.arrayToLinkedList([1,2,3,4,5])
print(result.next.data)
```
