set row and column to zero if there is zero in that row and column
- TC O((n*m))
- SC O(n+m)

```python
a = [[1,2,3,4],[4,5,10,6],[7,0,8,9],[1,2,3,4]]
k = {}
l = {}
for i in range(len(a)):
    for j in range(len(a[0])):
        if a[i][j] == 0:
            k[i]=0
            l[j]=0

for i in range(len(a)):
    for j in range(len(a[0])):
        if i in k or j in l:
            a[i][j] = 0
print(a)
```

another approch
- TC O((n*m))
- SC O(1)

| Step    | Purpose                                    | What happens                                |
| ------- | ------------------------------------------ | ------------------------------------------- |
| 1️⃣     | Detect if first row/col needs zeroing      | Save info in flags                          |
| 2️⃣     | Mark rows & cols that must become zero     | Store 0 in `matrix[i][0]` or `matrix[0][j]` |
| 3️⃣     | Use those markers to update rest of matrix | If any marker = 0 → cell = 0                |
| 4️⃣–5️⃣ | Finally, zero out first row/col if needed  | Based on earlier flags                      |


```python
a = [[1,2,3,4],[4,5,10,6],[7,0,8,9],[1,2,3,4]]
rows = len(a)
cols = len(a[0])
first_row_zeros = False
first_col_zeros = False
if any(a[0][j] == 0 for j in range(cols)):
    first_row_zeros = True
if any(a[i][0] == 0 for i in range(rows)):
    first_col_zeros = True
for i in range(1, rows):
    for j in range(1, cols):
        if a[i][j] == 0:
            a[i][0] = 0
            a[0][j] = 0

for i in range(1, rows):
    for j in range(1, cols):
        if a[i][0] == 0 or a[0][j] == 0:
            a[i][j] = 0

if first_row_zeros:
    for j in range(cols):
        a[0][j] = 0

if first_col_zeros:
    for i in range(rows):
        a[i][0] = 0

print(a)
```
