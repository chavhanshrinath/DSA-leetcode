Find the largest element in an array

```python
arr = [5,7,3,4,1,9,110,2,6,8]

largest = arr[0]

for i in arr:
	if i > largest:
		largest = i
print(largest)
```

Find the second largest number of array

```python
arr = [5,711,3,4,1,99,110,2,6,8]

largest, second_largest = float('-inf'), float('-inf')
# largest, second_largest = arr[0],arr[0]

for i in arr:
    if i > largest:
        second_largest = largest
        largest = i
    elif i > second_largest and i != largest:
        second_largest = i
print(second_largest)
print(largest, second_largest)
```

another approach

```python
def ckeck_sorted(arr):
    asc = True
    desc = True

    for i in range(len(arr)-1):
        if arr[i] < arr[i+1]:
            desc = False
        if arr[i] > arr[i+1]:
            asc = False
    
    if asc:
        return "Array is sorted in ascending order"
    elif desc:
        return "Array is sorted in descending order"
    else:
        return "Array is not sorted"

arr = [9,8,7,6,5,4,3,2,1]
print(ckeck_sorted(arr))
```

Remove the duplicates from sorted array [ Inplace ]
- TC O(n)
- SC O(1)

```python

a = [1,1,1,2,3,4,4,5,6]

i = 0
j = i+1

while j< len(a):
	if a[i] != a[j]:
		i+=1
		a[i],a[j] = a[j],a[i]
	j+=1
	
print(f"number of unique numbers--",i+1)
print(a)
```

Rotate the array to right by one place - inplace operation

```python
a = [2,5,4,73,3,5,6]
n = len(a)
temp = a[-1]
for i in range(n-2,-1,-1):
    a[i+1] = a[i]
a[0] = temp
print(a)
```

Rotate k elements of array - inplace operation

```python
a =[73, 3, 5, 6, 2, 5, 4]
print(id(a))
k = 3
n = len(a)
b = a[-k:]
for i in range(n-k-1,-1,-1):
    a[i+k] = a[i]
a[:k] = b
print(a)
print(id(a))

# here overall TC is O(n) and SC is O(k)



####### if sliding window is more than the array size then we can take mod of window size with array size ######

a = [1,2,3,4,5]
k = 12
n = len(a)
k = k % n

for _ in range(k):
    x = a.pop()
    a.insert(0,x)
print(a)

# here for loop and insert takes O(k) and O(n) time respectively so overall TC is O(n*k)
```

Optimal way to rotate the array and reverse the subarray 
- TC - O(n)

```python
def rotate_subarray(arr, start, end):
    # start  = 0
    # end  = len(arr)-1
    while start < end:
        arr[start], arr[end] = arr[end], arr[start]
        start += 1
        end -= 1
arr = [73, 3, 5, 6, 2, 5, 4]
n = len(arr)
k = 3
rotate_subarray(arr, n-k, n-1)
print(arr)
rotate_subarray(arr, 0, n-k-1)
print(arr)
rotate_subarray(arr, 0, n-1)
print(arr)
```

Move zeros to end

```python
a = [1,0,5,6,0,0,7,0,4]
n = len(a)
count = 0
for _ in range(n):      # iterate n times
    if 0 in a:          # time complexity O(n)
        a.remove(0)     # time complexity O(n)
        a.append(0)     # time complexity O(1)
    count+=1
print(a)
print(count)                # total time complexity is O(n^2)
```

another way  -  two pointer approach
- TC = O(n)

```python
a = [1,0,5,6,0,0,7,0,4]
i = 0
j = i+1
n = len(a)
while j < n:
    if a[i] == 0 and a[j] != 0:
        a[i], a[j] = a[j], a[i]
        i+=1
    if a[i] != 0:
        i+=1
    j+=1
print(a)
```

Merge 2 Sorted Arrays Without Duplicates *********IMP*******
- TC O(n+m)
- SC O(1)  / O(n+m)

```python

a = [1,1,1,3,4,5,6,7]
b = [1,2,3,6,7,89,10]
result = []
i,j =0,0

while i<len(a) and j< len(b):
	if a[i] <= b[j]:
		if len(result) == 0 or result[-1] != a[i]:
			result.append(a[i])
		i+=1
	else:
		if len(result) == 0 or result[-1] != b[j]:
			result.append(b[j])
		j+=1
		
while i< len(a):
	if len(result) == 0 or result[-1] != a[i]:
			result.append(a[i])
	i+=1

while j< len(b):
	if len(result) == 0 or result[-1] != b[j]:
			result.append(b[j])
	j+=1
	
print(result)
```

find the length of max consecutive ones

```python
a = [1,1,1,0,1,1,1,1,0,1,1,1,1,1,1]
max_count = 0
count =0
for i in a:
    if i == 1:
        count+=1
    else:
        max_count = max(max_count,count)
        count=0

print(max(max_count,count))
```
