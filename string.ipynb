 Remove Outermost Parentheses

```python

a = '(()())(())'
result = ''
count = 0
for i in a:
	if i =='(':
		count+=1
		if count >1:
			result = result+i
	else:
		count -=1
		if count !=0:
			result = result+i
print(result)
```

 Largest Odd Number in String

```python
class Solution(object):
    def largestOddNumber(self, num):
        """
        :type num: str
        :rtype: str
        """
        n = len(num)
        for i in range(n-1,-1,-1):
            if int(num[i])%2!=0:
                return num[:i+1]
                break
        return ""
        
sol = Solution()
result = sol.largestOddNumber("35427")
print(result)
```

```python
class Solution(object):
    def longestCommonPrefix(self, strs):
        """
        :type strs: List[str]
        :rtype: str
        """
        result = ''
        count = 0     
        base = strs[0]
        if len(strs)==1:
            return result + base
        for ch in range(len(base)):
            for i,j in enumerate(strs):
                if j.startswith(base[:ch]):
                    if count == i:
                        result = ''
                        result = result + base[:ch]
            count +=1
        return result 
sol = Solution()
result = sol.longestCommonPrefix(["a"])
print(result)
```

 longest Common Prefix --HARD

```python
def longestCommonPrefix(strs):
    if not strs:
        return ""
    
    prefix = strs[0]
    
    for s in strs[1:]:
        while not s.startswith(prefix):
            prefix = prefix[:-1]   # shorten prefix
            if not prefix:
                return ""
    return prefix

strs = ["flower","flor","flower","flower"]
print(longestCommonPrefix(strs))  # Output: "flower"
```
