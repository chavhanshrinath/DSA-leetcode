 Remove Outermost Parentheses

```python

a = '(()())(())'
result = ''
count = 0
for i in a:
	if i =='(':
		count+=1
		if count >1:
			result = result+i
	else:
		count -=1
		if count !=0:
			result = result+i
print(result)
```

```python
class Solution(object):
    def largestOddNumber(self, num):
        """
        :type num: str
        :rtype: str
        """
        n = len(num)
        for i in range(n-1,-1,-1):
            if int(num[i])%2!=0:
                return num[:i+1]
                break
        return ""
        
sol = Solution()
result = sol.largestOddNumber("35427")
print(result)
```
